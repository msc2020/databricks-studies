## Parsing with `ai_parse_document()` and Python

```python
from pyspark.sql.functions import expr

INPUT_DOCS_PATH = '/Volumes/studies/databricks_rag/docs_plano_diretor_tic'
all_docs = spark.read.format('binaryFile').load(INPUT_DOCS_PATH)

# parse docs
parsed_docs = all_docs.withColumn(
	'parsed_content',
	expr(
		f'''ai_parse_document(
			content, 
			map(
				'version', '2.0',
				'imageOutputPath', '{INPUT_DOCS_PATH}/parsed_images'
			)
		)'''
	)
)
```

## Parsing with `ai_parse_document()` and SQL

```python
parsed_docs_sql = spark.sql(f'''
select
	path,
	ai_parse_document(
		content,
		map('version', '2.0')
	) as parsed_docs
from read_files('{INPUT_DOCS_PATH}', format => 'binaryFile')
''')
```

```python
display(parsed_docs_sql)
```

## Checking the metadata of parsed documents

```python
from pyspark.sql.functions import expr

metadata_docs_parsed = parsed_docs.select(
	'path',
	expr('parsed_content:document:pages'),
	expr('parsed_content:document:elements'),
	expr('parsed_content:document:error_status'),
	expr('parsed_content:document:corrupted_data'),
	expr('parsed_content:document:metadata')
)

#  column, variable, or function parameter with name `parsed_docs` cannot be resolved. Did you mean one of the following? [`parsed_content`, `length`, `path`, `content`, `modificationTime`].
```
