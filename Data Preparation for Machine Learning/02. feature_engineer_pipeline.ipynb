```python
cols = [
    'Hectares ', 'Agriblock', 'Variety', 'Soil Types', 'Nursery',
    'Nursery area (Cents)', 'Trash(in bundles)', 'LP_nurseryarea(in Tonnes)', 'LP_Mainfield(in Tonnes)',
    'Paddy yield(in Kg)'
]

df_paddy_selected = df_paddy.select(cols)
df_paddy_selected.count()
```

```python
df_train, df_test = df_paddy_selected.randomSplit([0.7, 0.3], seed=42)
display(df_train.summary())
display(df_test.summary())
```

### Convert int and bool columns todouble

```python
target_col = 'Paddy yield(in Kg)'

numeric_types = (IntegerType, BooleanType)

integer_cols = [
    f.name
    for f in df_train.schema.fields
    if isinstance(f.dataType, numeric_types)
    and f.name != target_col
]
```

```python
for col_i in integer_cols:
    df_train = df_train.withColumn(col_i, col(col_i).cast(DoubleType()))
    df_test = df_test.withColumn(col_i, col(col_i).cast(DoubleType()))
```

### Missing Value Treatment

```python
num_cols = [c.name for c in df_train.schema.fields if c.dataType == DoubleType()]

num_missing_values = [count(when(col(col_i).isNull(), col_i)).alias(col_i) for col_i in num_cols]
row_dict_num = df_train.select(num_missing_values).collect()[0].asDict()
num_missing_cols = [col_i for col_i in num_cols if row_dict_num[col_i] > 0]

print(f'Numeric columns with misssing values: {num_missing_cols}')
```

```python

str_cols = [c.name for c in df_train.schema.fields if c.dataType == DoubleType()]

str_missing_values = [count(when(col(col_i).isNull(), col_i)).alias(col_i) for col_i in str_cols]
row_dict_str = df_train.select(str_missing_values).collect()[0].asDict()
str_missing_cols = [col_i for col_i in num_cols if row_dict_str[col_i] > 0]

print(f'String columns with misssing values: {str_missing_cols}')
```

## Pipeline

### Encode

```python
display(df_paddy_selected.limit(5))
```

```python
print(df_paddy_selected.columns)
```

```python
cat_cols = [
    'Hectares ', 'Agriblock', 'Variety', 'Soil Types', 'Nursery'
]

cat_cols_indexed = [f'{col_i}_index' for col_i in cat_cols]
ohe_cols = [f'{col_i}_ohe' for col_i in cat_cols]
ohe_cols
```

```python
indexer = StringIndexer(
    inputCols=cat_cols,
    outputCols=cat_cols_indexed,
    handleInvalid='keep'
)

ohe = OneHotEncoder(
    inputCols=cat_cols_indexed,
    outputCols=ohe_cols,
    handleInvalid='keep'
)
```

### Assemble and Scale

```python
num_cols
```

```python
num_assembler = VectorAssembler(
    inputCols=num_cols,
    outputCol='num_vec_assembled'
)

scaler = StandardScaler(
    inputCol='num_vec_assembled',
    outputCol='num_scaled'
)
```

### Assemble all features into a single vector



```python
feature_cols = ['num_scaled'] + ohe_cols

vector_assembler = VectorAssembler(
    inputCols=feature_cols,
    outputCol='all_features'
)
```

### Define the pipeline

```python
df_train.dtypes
```

```python
stages_list = [indexer, num_assembler, scaler, ohe, vector_assembler]

pipeline = Pipeline(stages=stages_list)
```

### Fit the pipeline

```python
pipe_model = pipeline.fit(df_train)
```

### Transform with pipeline

```python
df_train_transformed = pipe_model.transform(df_train)
df_train_transformed.select('all_features').show(5, truncate=False)
```

```python
df_test_transformed = pipe_model.transform(df_test)
```

```python
display(df_test_transformed.select('all_features', target_col))
```

### Export the pipeline

```python
output_model_path = '/Volumes/studies/data_prep_ml/volume-datalake/pipeline_model'
pipe_model.write().save(output_model_path)
output_model_path
```
