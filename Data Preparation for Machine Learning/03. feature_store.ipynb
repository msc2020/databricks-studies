# Data Preparation for Machine Learning

## Initial config

```python
data_path = r'/Volumes/studies/data_prep_ml/volume-datalake/paddydataset.csv'           

df_paddy = spark.read.option('NullValue', ' ').csv(
  data_path,
  header=True,
  inferSchema=True,
  multiLine=True,
  escape='"'
)

df_paddy_pk = df_paddy.withColumn('Id', F.monotonically_increasing_id())
```

```python
display(df_paddy_pk.limit(5))
```

```python
print(df_paddy_pk.columns)
```

```python
cols = [
    'Hectares ', 'Agriblock', 'Variety', 'Soil Types', 'Nursery',
    'Nursery area (Cents)', 'Trash(in bundles)', 'LP_nurseryarea(in Tonnes)', 'LP_Mainfield(in Tonnes)',
    'Paddy yield(in Kg)', 'Id'
]

df_paddy_selected = df_paddy_pk.select(cols)
df_paddy_selected.count()
```

### Feature store

```python
fs_client = FeatureStoreClient()
```

```python
def clean_column(col_name):
    col_name = col_name.strip()
    col_name = col_name.lower()
    col_name = re.sub(r"[ ,;{}()\n\t=]", "_", col_name)
    col_name = re.sub(r"_+", "_", col_name)
    return col_name

df_paddy_clean = df_paddy_pk.toDF(*[clean_column(c) for c in df_paddy_pk.columns])
```

```python
display(df_paddy_clean.limit(5))
```

```python
table_name = f'studies.data_prep_ml.paddy_features'

fs_client.create_table(
    name=table_name,
    primary_keys=['id'],
    df=df_paddy_clean,
    description='Paddy dataset | source=bronze | format=delta',
)
```

```python
feat_table = fs_client.get_table(name=table_name)
print(f'Description: {feat_table.description}')
print(f'\nFeatures: {feat_table.features}')
```

```python
display(fs_client.read_table(name=table_name))
```

```python
display(df_paddy_clean.summary())
```

```python
df_paddy_updated = df_paddy_clean.withColumn(
	'dap_group',
 	when((df_paddy_clean.dap_20days >= 0) & (df_paddy_clean.dap_20days <= 40), 'low')
  	.when((df_paddy_clean.dap_20days > 40) & (df_paddy_clean.dap_20days <= 120), 'mid')
 	.when((df_paddy_clean.dap_20days > 120) & (df_paddy_clean.dap_20days <= 200), 'high')
  	.when((df_paddy_clean.dap_20days > 200) & (df_paddy_clean.dap_20days <= 240), 'very_high')	
	.otherwise('invalid')
)
```

```python
fs_client.write_table(
	name=table_name,
	df=df_paddy_updated.select('id', 'dap_group'),  # primary key and column to add
	mode='merge'
)
```

```python
%sql
alter table paddy_features set tblproperties (
	'delta.columnMapping.mode' = 'name',
	'delta.minReaderVersion' = '2',
	'delta.minWriterVersion' = '5'
);
```

```python
timestamp_v3 = spark.sql(f'describe history {table_name}').orderBy('version').collect()[2].timestamp
print(timestamp_v3)
```

```python
df_v3 = spark.read.option('timestampAsOf', timestamp_v3).table(table_name)

display(df_v3)
```

```python
feature_df = fs_client.read_table(
	name=table_name,
	as_of_delta_timestamp=timestamp_v3
)

feature_df.printSchema()
```
